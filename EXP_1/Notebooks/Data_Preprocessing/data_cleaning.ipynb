# CREATING CLEANED DATA BY EACH JOINTS

```python
import os
import pandas as pd
```

```python
folder_path = r"S:/Nit Durgapur/College 6th Sem/CSIR-CMERI/Friction_Modelling/EXP_1/EXP_1_Data/12_06_2026/Actual_Data"

joint_files = {
    1: "J_1_12.06.2026_13_05_36.csv",
    2: "J_2_12.06.2026_13_22_35.csv",
    3: "J_3_12.06.2026_13_13_15.csv",
    5: "J_5_12.06.2026_13_14_50.csv"
}
```

```python
for joint_no, filename in joint_files.items():
    # Read original file
    file_path = os.path.join(folder_path, filename)
    df = pd.read_csv(file_path)
    # Keep only the corresponding joint columns
    clean_df = df[
        [
            'timestamp',
            f'm_t_{joint_no}',
            f'jts{joint_no}',
            f'v{joint_no}',
            f'q{joint_no}'
        ]
    ].copy()
    # Save with original column names
    output_file = os.path.join(folder_path, f'Joint_{joint_no}_Repetitive.csv')
    clean_df.to_csv(output_file, index=False)
    print(f"Saved: {output_file}")
```

# CREATING UNIQUE DATA FOR ACCELERATION

```python
folder_path = r"S:/Nit Durgapur/College 6th Sem/CSIR-CMERI/Friction_Modelling/EXP_1/EXP_1_Data/12_06_2026/Cleaned_Data_With_Acc"

joint_files = {
    1: "1_Acc.csv",
    2: "2_Acc.csv",
    3: "3_Acc.csv",
    5: "5_Acc.csv"
}

folder_path_acc = r"S:/Nit Durgapur/College 6th Sem/CSIR-CMERI/Friction_Modelling/EXP_1/EXP_1_Data/12_06_2026/Unique_Data_With_Acc"

# Iterate over each joint and its corresponding file
for joint, file_name in joint_files.items():
    # Construct the full path to the input file
    full_file_path = os.path.join(folder_path, file_name)
    
    # Check if the file exists before attempting to read
    if os.path.exists(full_file_path):
        # Read the CSV
        df = pd.read_csv(full_file_path)
        
        # 1. Identify and remove time series columns
        time_cols = ['timestamp', 'time_seconds']
        cols_to_drop = [col for col in time_cols if col in df.columns]
        
        if cols_to_drop:
            df.drop(columns=cols_to_drop, inplace=True)
            print(f"[{file_name}] Dropped columns: {cols_to_drop}")
        else:
            print(f"[{file_name}] No time series columns found to drop.")

        # 2. Remove duplicate rows
        # This checks all remaining columns and keeps only the first unique instance
        initial_rows = len(df)
        df_unique = df.drop_duplicates(keep='first').reset_index(drop=True)
        final_rows = len(df_unique)

        # Output stats
        print("\n" + "=" * 40)
        print(f"File Processed : {file_name} (Joint {joint})")
        print(f"Original rows : {initial_rows}")
        print(f"Unique rows remaining : {final_rows}")
        print(f"Duplicate rows removed : {initial_rows - final_rows}")
        print("=" * 40 + "\n")

        # 3. Save the cleaned data to a new file in the same folder
        output_name = f"{joint}_Acc_Unique.csv"
        output_file_path = os.path.join(folder_path_acc, output_name)
        
        df_unique.to_csv(output_file_path, index=False)
        print(f"Successfully saved to: {output_file_path}\n")

    else:
        print(f"Error: The file '{full_file_path}' was not found in the directory.")

print("Data processing complete.")
```
