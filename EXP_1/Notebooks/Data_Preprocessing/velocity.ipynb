```python
import pandas as pd
import os

folder_path = r"S:/Nit Durgapur/College 6th Sem/CSIR-CMERI/Friction_Modelling/EXP_1/EXP_1_Data/12_06_2026/Cleaned_Data_By_Joints"

files = {
    1: "Joint_1_Repetitive.csv",
    2: "Joint_2_Repetitive.csv",
    3: "Joint_3_Repetitive.csv",
    5: "Joint_5_Repetitive.csv"
}

folder_path_vel = r"S:/Nit Durgapur/College 6th Sem/CSIR-CMERI/Friction_Modelling/EXP_1/EXP_1_Data/12_06_2026/Cleaned_Data_With_Vel"

# Iterate through both the key (joint_num) and value (filename)
for joint_num, filename in files.items():
    try:
        # 1. Load the dataset
        file_path = os.path.join(folder_path, filename)
        df = pd.read_csv(file_path)
        
        # We only need to reference the position column dynamically now
        q_col = f'q{joint_num}'
        
        # 2. Convert timestamp to datetime objects
        df['time_obj'] = pd.to_datetime(df['timestamp'], format='%H:%M:%S.%f')
        
        # 3. Create the 'time_seconds' column (elapsed time starting from 0.0s)
        df['time_seconds'] = (df['time_obj'] - df['time_obj'].iloc[0]).dt.total_seconds()
        
        # 4. Calculate del_t using our new time_seconds column
        df['del_t'] = df['time_seconds'].diff()
        
        # 5. Calculate true velocity (del_q / del_t)
        df['del_q'] = df[q_col].diff()
        df['velocity'] = df['del_q'] / df['del_t']
        
        # 6. Clean up the dataframe: Keep ALL original columns, plus the new ones.
        df_final = df.drop(columns=['time_obj', 'del_t', 'del_q'])

        # 7. Save to CSV in the specified folder path
        output_filename = f'{joint_num}_Vel.csv'
        output_path = os.path.join(folder_path_vel, output_filename)
        df_final.to_csv(output_path, index=False)
        
        print(f"Successfully processed {filename} -> Saved to {output_filename}")
        
    except Exception as e:
        print(f"Error processing {filename}: {e}")
```
