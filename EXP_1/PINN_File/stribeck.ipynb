```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.preprocessing import StandardScaler

import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping
```

```python
N = 100 # Gear_Ratio

folder_path = r"S:/Nit Durgapur/College 6th Sem/CSIR-CMERI/Friction_Modelling/EXP_1/EXP_1_Data/12_06_2026"
file = r"J_1_12.06.2026_13_05_36.csv"

file_path = os.path.join(folder_path, file)
df = pd.read_csv(file_path)
print(df.shape)
df.head()
```

```python
df["q"] = np.deg2rad(df["q1"])
df["v"] = np.deg2rad(df["v1"])
df["tm"] = df["m_t_1"]
df["tj"] = df["jts1"]
```

```python
df["tf"] = (N * df["tm"]) - df["tj"]
```

```python
size = len(df)

train_end = int(0.60 * size)
val_end = int(0.80 * size)

train_df = df.iloc[: train_end]
val_df = df.iloc[train_end : val_end]
test_df = df.iloc[val_end :]

print(train_df.shape)
print(val_df.shape)
print(test_df.shape)
```

```python
def model1(v, Tc, Bv, Vs):
    return (Tc * np.tanh(v/Vs) + Bv * v)
```

```python
v_train = train_df["v"].values
tf_train = train_df["tf"].values

p0 = [5, 0.1, 0.01]

params1, _ = curve_fit(
    model1,
    v_train,
    tf_train,
    p0 = p0,
    maxfev = 50000
)

Tc1, Bv1, Vs1 = params1

print("Tc =",Tc1)
print("Bv =",Bv1)
print("Vs =",Vs1)
```

```python
v_test = test_df["v"].values
tf_test = test_df["tf"].values

pred1 = model1(v_test, * params1)
```

```python
rmse1 = np.sqrt(mean_squared_error(tf_test, pred1))
mae1 = mean_absolute_error(tf_test, pred1)
r2_1 = r2_score(tf_test, pred1)

print(rmse1)
print(mae1)
print(r2_1)
```

```python
def model2(v, Ts, Tc, Bv, Vs, alpha):
    return ((Tc + (Ts-Tc) * np.exp(-(np.abs(v)/Vs) ** alpha)) * np.tanh(v/0.001) + Bv * v)
```

```python
p0 = [10, 5, 0.1, 0.05, 1]

params2,_ = curve_fit(
    model2,
    v_train,
    tf_train,
    p0=p0,
    maxfev=100000
)

Ts, Tc, Bv, Vs, alpha = params2

print("Tc = ", Tc)
print("Ts = ", Ts)
print("Bv = ", Bv)
print("Vs = ", Vs)
print("alpha = ", alpha)
```

```python
pred2 = model2(v_test, * params2)
```

```python
rmse2 = np.sqrt(mean_squared_error(tf_test, pred2))
mae2 = mean_absolute_error(tf_test, pred2)
r2_2 = r2_score(tf_test, pred2)

print(rmse2)
print(mae2)
print(r2_2)
```

```python
plt.figure(figsize=(15,5))
plt.plot(
    tf_test,
    label="Actual Torque",
    linewidth=1
)
plt.plot(
    pred2,
    label="Stribeck Prediction",
    linewidth=1
)
plt.xlabel("Sample Number")
plt.ylabel("Torque")
plt.title(f"Stribeck Model\nR²={r2_2:.4f}, RMSE={rmse2:.4f}")
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(7,7))
plt.scatter(
    tf_test,
    pred2,
    s=5,
    alpha=0.3
)
min_val = min(
    tf_test.min(),
    pred2.min()
)
max_val = max(
    tf_test.max(),
    pred2.max()
)
plt.plot(
    [min_val,max_val],
    [min_val,max_val],
    'r',
    linewidth=2
)
plt.xlabel("Actual Torque")
plt.ylabel("Predicted Torque")
plt.title("Stribeck Model\nActual vs Predicted")
plt.grid(True)
plt.show()
```

```python
idx = np.argsort(v_test)
plt.figure(figsize=(10,6))
plt.scatter(
    v_test,
    tf_test,
    s=2,
    alpha=0.2,
    label="Measured"
)
plt.plot(
    v_test[idx],
    pred2[idx],
    linewidth=3,
    color='red',
    label='Stribeck Model'
)
plt.xlabel("Velocity (rad/s)")
plt.ylabel("Residual Torque")
plt.title("Stribeck Friction Curve")
plt.legend()
plt.grid(True)
plt.show()
```

```python
residual2 = tf_test - pred2
plt.figure(figsize=(15,4))
plt.plot(
    residual2,
    linewidth=1
)
plt.axhline(0, color='red')
plt.xlabel("Sample Number")
plt.ylabel("Residual Error")
plt.title("Stribeck Residual Error")
plt.grid(True)
plt.show()
```
